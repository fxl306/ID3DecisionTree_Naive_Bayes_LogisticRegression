# **Question(c)**
For volcanoes and spam, plot the CV accuracy as the depth of the tree is increased. On the x-axis, choose depth values 1, 2, 4, 8, 16 and 32. Explain the pattern of the graph?

````python
import matplotlib.pyplot as plt
````

- **Volcanoes Dataset**

````python
Accuracy = [0.67, 0.752, 0.778, 0.794, 0.776, 0.776]
x = [1, 2, 4, 8, 16, 32]
plt.plot(x, Accuracy, color="blue", label="Accuracy")
plt.xlabel("Depth")
plt.ylabel("Accuracy")
plt.xticks(x)
plt.yticks([0.5,0.75,1])
plt.title("Volcaones Dataset")
plt.legend()
````

Above plot represents the accuracy of the volcanoes dataset , it increase as the depth level increases, when the depth increases to  8, the change of the accuracy becomes small if continues to increase. This because the volcanoes dataset is relatively small, increasing the depth level helps to improve the performance of the decision tree, and will become stable when the depth reaches to a certain level, such as depth = 8. 

- **Spam Dataset**

````python
Accuracy = [0.67, 0.73, 0.75, 0.79, 0.81, 0.81]
x = [1, 2, 4, 8, 16, 32]
plt.plot(x, Accuracy, color="blue", label="Accuracy")
plt.xlabel("Depth")
plt.ylabel("Accuracy")
plt.xticks(x)
plt.yticks([0,0.5,0.8,1])
plt.title("Spam Dataset")
plt.legend()
````

Above plot represents the accuracy of the “spam” dataset , it increases as the depth level increases, and becomes stable when the depth reaches to 8. "Spam” dataset is a larger dataset than the volcanoes datasets, increasing the depth level helps to improve the performance of the decision tree, and will become stable when the depth reaches to a certain level, such as depth = 8. Thus also demostrate the good performance our algorithm.
